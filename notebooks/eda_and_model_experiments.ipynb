This notebook documents the research, data analysis, feature engineering, and model experimentation for the **Hybrid Movie Recommendation Engine**.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.decomposition import TruncatedSVD

# Set dark plotting style
plt.style.use('dark_background')
%matplotlib inline
```

## 1. Load Processed Datasets

```python
data_dir = os.path.join("..", "data", "processed")
movies_df = pd.read_csv(os.path.join(data_dir, "movies_processed.csv"))
ratings_df = pd.read_csv(os.path.join(data_dir, "ratings_processed.csv"))

print(f"Movies Dataset Shape: {movies_df.shape}")
print(f"Ratings Dataset Shape: {ratings_df.shape}")
display(movies_df.head(5))
```

## 2. Exploratory Data Analysis (EDA)

```python
# Rating Distribution
plt.figure(figsize=(10, 5))
sns.histplot(ratings_df['rating'], bins=10, kde=True, color='#E50914')
plt.title('MovieLens Ratings Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Rating (1-5 Stars)')
plt.ylabel('Frequency')
plt.grid(alpha=0.2)
plt.show()
```

```python
# Top 15 Genres Frequency
all_genres = []
for g_str in movies_df['genres'].dropna():
    if g_str != '(no genres listed)':
        all_genres.extend(g_str.split('|'))

genre_counts = pd.Series(all_genres).value_counts().head(15)

plt.figure(figsize=(12, 6))
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='viridis')
plt.title('Top 15 Movie Genres Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Number of Movies')
plt.grid(alpha=0.2)
plt.show()
```

## 3. TF-IDF & Cosine Similarity Matrix

```python
tfidf = TfidfVectorizer(max_features=1000, stop_words='english')
tfidf_matrix = tfidf.fit_transform(movies_df['corpus'].fillna(''))
print(f"TF-IDF Matrix Shape: {tfidf_matrix.shape}")

# Compute Cosine Similarity subset sample
sample_sim = cosine_similarity(tfidf_matrix[:50], tfidf_matrix[:50])

plt.figure(figsize=(10, 8))
sns.heatmap(sample_sim, cmap='magma')
plt.title('Sample Cosine Similarity Matrix Heatmap (First 50 Movies)', fontsize=14, fontweight='bold')
plt.show()
```

## 4. Collaborative SVD Latent Factor Analysis

```python
# User-Item matrix dimensions & sparsity analysis
num_users = ratings_df['userId'].nunique()
num_movies = ratings_df['movieId'].nunique()
sparsity = 1.0 - (len(ratings_df) / (num_users * num_movies))

print(f"Unique Users: {num_users}")
print(f"Unique Movies: {num_movies}")
print(f"Rating Matrix Sparsity: {sparsity * 100:.2f}%")
```
